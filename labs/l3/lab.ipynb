# Lab 3: Major Vote Classifier

- find the most common label
- predict that label
- calculate error rate for predictions
- output class label is binary
- output: higher, descending alphabetically

```python
import pandas as pd
```

## Heart Dataset

### Prepare the dataset

Load dataset

```python
heart_tr = pd.read_csv("heart_train.tsv", sep = "\t")
heart_te = pd.read_csv("heart_test.tsv", sep = "\t")
```

```python
X_tr = heart_tr.iloc[:, :-1]
y_tr = heart_tr.iloc[:, -1]

X_te = heart_te.iloc[:, :-1]
y_te = heart_te.iloc[:, -1]
```

```python
def printModel(model: dict):
	print("Printing model:")
	print("-" * 70)
	for c in model:
		print(c.split("_")[0][:4], end="\t")
	print()
	for c in model:
		print(model[c], end="\t")
	print()
	print("-" * 70)
```

```python
# initialize model
model1 = {}
for c in heart_tr.columns[:-1]:
	model1[c] = 0

printModel(model1)
```

```python
printModel(model1)

# fit the model
for i in range(len(X_tr)):
	for c in X_tr.columns:
		feature = X_tr.loc[i, c]
		target = y_tr.loc[i]
		# print(i, c, X_tr.loc[i, c], y_tr.loc[i])
		if (feature == 0 and target == 0) or (feature == 1 and target == 1):
			model1[c] += 1

printModel(model1)
```

```python
# find the most common label
def getMostCommonLabel(model: dict):
	return max(model.items(), key = lambda k: (k[1], k[0]))[0]
```

```python
def getErrorRate(model: dict, X, y, n=10):
	commonLabel = getMostCommonLabel(model)
	print("Most common label:", commonLabel)

	error = 0
	for i in range(len(X)):
		feature = X.loc[i, commonLabel]
		target = y.loc[i]
		if i <= n:
			print(f"{i}: feature[{commonLabel}]={feature}, target={target}, match={feature == target}")
		if feature != target:
			error += 1

	return error / len(X)



error_tr = getErrorRate(model1, X_tr, y_tr)
error_te = getErrorRate(model1, X_te, y_te)

print("Error rate on training set:", error_tr)
print("Error rate on test set:", error_te)
```
