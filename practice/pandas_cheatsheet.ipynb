## 4. Deal with empty values

First let's deal with the empty values in the Item Column, since we have a "Price Per Unit" we can deduce what item that was

````python
#let's make a price dict
item_price = {
    'Coffee': 2.0,
    'Tea': 1.5,
    'Sandwich': 4.0,
    'Salad': 3.0,
    'Cake': 3.0,
    'Cookie': 1.0,
    'Smoothie': 4.0,
    'Juice': 3.0
}

# Reverse mapping (price to item)
price_item = {price: item for item, price in item_price.items()}

print(item_price)
print(price_item)
````

````python
try:
    df['Price Per Unit']=df['Price Per Unit'].astype(float)
except Exception as e:
    print(e)
    pass
````

### 1. Detect Null values

````python
df.isnull().sum()
````

````python
df.isna().sum()
````

### 2. Convert strings to Null values

Can't perform _fillna()_ if NA is a string

```python
numpy.replace([x, y], np.nan)
```

````python
import numpy as np
df['Price Per Unit'] = df['Price Per Unit'].replace(['ERROR', 'UNKNOWN'], np.nan)
df['Price Per Unit'] = df['Price Per Unit'].astype(float)
df["Price Per Unit"].info()
````

````python
df['Item'] = df['Item'].fillna(df['Price Per Unit'].map(price_item))
df.info()
````

#### Converting Total Spent

````python
df['Total Spent'] = df['Total Spent'].replace(['ERROR', 'UNKNOWN'], np.nan)
df['Total Spent'] = df['Total Spent'].astype(float)
````

#### Converting Quantity

````python
df['Quantity'] = df['Quantity'].replace(['ERROR', 'UNKNOWN'], np.nan)
df['Quantity']=df['Quantity'].astype(float)
````

### 3. Filling NA values

````python
df['Price Per Unit'] = df['Price Per Unit'].fillna(df['Total Spent'] / df['Quantity'])
df.info()
````

#### Fillna _Item_ column

````python
df['Item'] = df['Item'].fillna(df['Price Per Unit'].map(price_item))
df.info()
````

### 4. Drop NA rows

If remaining NA values don't make sense

````python
df.dropna(subset=['Item'], inplace=True)
df.info()
````

#### Dropping _Transaction Date_ NA rows

````python
df.dropna(subset=['Transaction Date'], inplace=True)
df.info()
````
